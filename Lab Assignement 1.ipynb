```python

students = []
def calculate_result(total):
    percentage = total / 300 * 100

    if percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"
    return percentage, grade
for i in range(5):
    print("\nStudent", i + 1)
    name = input("Enter name: ")
    roll = input("Enter roll number: ")
    p = float(input("Python marks: "))
    ai = float(input("AI marks: "))
    math = float(input("Mathematics marks: "))
    total = p + ai + math
    percentage, grade = calculate_result(total)

    student = {
        "name": name,
        "roll": roll,
        "total": total,
        "percentage": percentage,
        "grade": grade
    }
    students.append(student)
highest = students[0]
file = open("results.txt", "w")
for s in students:
    print("\nName:", s["name"])
    print("Roll Number:", s["roll"])
    print("Total:", s["total"])
    print("Percentage:", round(s["percentage"], 2))
    print("Grade:", s["grade"])
    file.write(str(s) + "\n")
    if s["percentage"] > highest["percentage"]:
        highest = s
print("\nHighest Percentage:", highest["name"])
file.write("\nHighest Percentage: " + highest["name"])
file.close()
print("Results saved successfully.")
```

```python
def analyze(sentence):
    words = sentence.split()
    print("Total words:", len(words))
    print("Total characters:", len(sentence))
    if len(words) > 0:
        longest = words[0]
        shortest = words[0]
        for word in words:
            if len(word) > len(longest):
                longest = word
            if len(word) < len(shortest):
               shortest = word
        print("Longest word:", longest)
        print("Shortest word:", shortest)
    vowels = 0
    for ch in sentence.lower():
        if ch in "aeiou":
            vowels += 1
    print("Total vowels:", vowels)
    unique_words = set()
    for word in words:
        unique_words.add(word.lower())
    print("Unique words:", unique_words)
    sorted_words = sorted(unique_words)
    print("Alphabetical order:", sorted_words)
    file = open("word_analysis.txt", "w")
    file.write("Sentence: " + sentence + "\n")
    file.write("Total words: " + str(len(words)) + "\n")
    file.write("Total characters: " + str(len(sentence)) + "\n")
    file.write("Longest word: " + (longest if words else "None") + "\n")
    file.write("Shortest word: " + (shortest if words else "None") + "\n")
    file.write("Vowels: " + str(vowels) + "\n")
    file.write("Unique words: " + str(sorted_words))
    file.close()
text = input("Enter a sentence: ")
analyze(text)
print("Analysis saved successfully.")
```

```python
products = {}
def calculate_bill(products):
    total = 0
    for name in products:
        price = products[name]["price"]
        quantity = products[name]["quantity"]
        amount = price * quantity
        products[name]["total"] = amount
        total = total + amount
        print(name, "Total: Rs.", amount)
    discount = 0
    if total > 10000:
        discount = total * 0.10
    final_bill = total - discount
    return total, discount, final_bill
for i in range(5):
    print("\nProduct", i + 1)
    name = input("Product name: ")
    price = float(input("Price: "))
    quantity = int(input("Quantity: "))
    products[name] = {
        "price": price,
        "quantity": quantity
    }
total, discount, final = calculate_bill(products)
expensive = list(products.keys())[0]
for name in products:
    if products[name]["price"] > products[expensive]["price"]:
        expensive = name
print("\nTotal Bill:", total)
print("Discount:", discount)
print("Final Bill:", final)
print("Most expensive product:", expensive)
file = open("bill.txt", "w")
file.write(str(products) + "\n")
file.write("Total Bill: " + str(total) + "\n")
file.write("Discount: " + str(discount) + "\n")
file.write("Final Bill: " + str(final))
file.close()
print("Bill saved successfully.")
```

```python
def add(a, b):
    return a + b
def subtract(a, b):
    return a - b
def multiply(a, b):
    return a * b
def divide(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b
def modulus(a, b):
    if b == 0:
        return "Cannot use zero"
    return a % b
while True:
    a = float(input("\nEnter first number: "))
    b = float(input("Enter second number: "))
    print("+ Addition")
    print("- Subtraction")
    print("* Multiplication")
    print("/ Division")
    print("% Modulus")
    op = input("Choose operation: ")
    if op == "+":
        result = add(a, b)
    elif op == "-":
        result = subtract(a, b)
    elif op == "*":
        result = multiply(a, b)
    elif op == "/":
        result = divide(a, b)
    elif op == "%":
        result = modulus(a, b)
    else:
        result = "Invalid operation"
    print("Result:", result)
    again = input("Calculate again? (yes/no): ")
    if again.lower() != "yes":
        break
print("Calculator closed.")
```

```python
menu = {
    1: ["Burger", 500],
    2: ["Pizza", 800],
    3: ["Fries", 250],
    4: ["Drink", 150]
}
orders = {}
def calculate_bill(orders):
    total = 0
    for item in orders:
        price = menu[item][1]
        quantity = orders[item]
        total = total + price * quantity
    return total
while True:
    print("\nRestaurant Menu")
    for item in menu:
        print(item, menu[item][0], "Rs.", menu[item][1])
    print("0 Exit")
    choice = int(input("Choose item: "))
    if choice == 0:
        break
    if choice in menu:
        quantity = int(input("Enter quantity: "))
        if quantity > 0:
            if choice in orders:
                orders[choice] = orders[choice] + quantity
            else:
                orders[choice] = quantity
        else:
            print("Invalid quantity")
    else:
        print("Invalid choice")
total = calculate_bill(orders)
file = open("orders.txt", "w")
file.write("Restaurant Bill\n")
for item in orders:
    name = menu[item][0]
    price = menu[item][1]
    quantity = orders[item]
    amount = price * quantity
    print(name, "x", quantity, "=", amount)
    file.write(name + " x " + str(quantity) + " = Rs. " + str(amount) + "\n")
print("Final Bill: Rs.", total)
file.write("Final Bill: Rs. " + str(total))
file.close()
print("Bill saved successfully.")
```
